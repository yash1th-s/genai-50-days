# Day 44: Run AI on Your Own Laptop

Every day so far, our code called Google's servers. Today — **nothing leaves your machine.**

We run a real LLM *locally* with **Ollama**, wire it into LangChain, and prove it: no internet, no API key, ₹0. Same code you already know — we just swap the model.

## Install

One Python package. The models run in **Ollama** — a free app that serves LLMs on your own computer (download once from ollama.com, then `ollama pull gemma3:1b`).

````python
%pip install langchain-ollama --quiet
````

## Setup — No API Key

Notice what's **missing**: no `load_dotenv`, no `GOOGLE_API_KEY`. Ollama runs on your laptop, so there's nothing to authenticate and nothing to pay.

````python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gemma3:1b", temperature=0)
print("✅ Model ready — running fully on this laptop, no key needed")
````

## Talk to It

A real language model answering from your own hardware — in a fraction of a second.

````python
import time

t = time.time()
reply = model.invoke("In one sentence, what is a large language model?")
print(reply.content.strip())
print(f"\n⌚ {time.time() - t:.1f}s — and that ran with your Wi-Fi off")
````

## It's Just LangChain — Same Code as Always

The whole point: **everything downstream is unchanged.** Prompts, chains, parsers — the exact patterns from the cloud days work on the local model.

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    "You are a travel guide. Suggest ONE thing to do in {city}. One short sentence."
)
chain = prompt | model | StrOutputParser()   # the same | pipe from the LangChain days

print("Paris →", chain.invoke({"city": "Paris"}).strip())
print("Tokyo →", chain.invoke({"city": "Tokyo"}).strip())
````

## The One-Line Swap

This is why LangChain matters: it's **model-agnostic.** Cloud or local, your code barely changes.

```python
# CLOUD (Days 26–43)
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# LOCAL (today) — swap two lines, everything else is identical
from langchain_ollama import ChatOllama
model = ChatOllama(model="gemma3:1b")
```

Need a different brain? `ChatOllama(model="qwen3:1.7b")` — or Llama, Mistral, Phi. Pull it once, swap the name.

## Why This Is a Big Deal

- **₹0 forever** — no per-token bill, no quota, run it a million times
- **Private** — your prompts and data never leave the machine (medical, legal, personal)
- **Offline** — works on a plane, in a basement, with the Wi-Fi off
- **Trade-off:** a 1B local model isn't as sharp as a frontier cloud model — pick local for privacy/cost/offline, cloud for raw power

## Key Takeaways

1. **You can run a real LLM locally** — Ollama + `langchain-ollama`, no API key, no internet, zero cost
2. **LangChain is model-agnostic** — swap `ChatGoogleGenerativeAI` → `ChatOllama` and every prompt, chain, and parser just works
3. **Local vs cloud is a trade-off** — local wins on privacy, cost, and offline; cloud wins on raw capability
